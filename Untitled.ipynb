```python
import pandas as pd
data = pd.read_csv("cooking_fuel.csv")
```

```python
data.rename(columns = {"size":"size(lb)"}, inplace = True)
```

```python
data.head()
```

```python
data["size(lb)"] = data["size(lb)"].str.replace("lb","",regex = False).astype(float)
```

```python
data.head()
```

```python
data["sale_date"] = pd.to_datetime(data["sale_date"])
```

```python
print(data["sale_date"].dtype)
```

```python
data["name"] = (data["first_name"].fillna("") + " " + data["last_name"].fillna("")).str.strip()
```

```python
data.head()
```

```python
data = data.drop(columns = ["first_name","last_name"])
```

```python
col = data.pop("name")
data.insert(2, "name" , col)
```

```python
data.head()
```

```python
invalid_emails = ~data["email"].str.match(r"^[\w\.-]+@[\w\.-]+\.\w+$", na=False)

invalid_count = invalid_emails.sum()

print(invalid_count)
```

```python
print(data.loc[invalid_emails, "email"])
```

```python
invalid_emails.head()
```

```python
good_e = ~data["email"].str.match(r"^[\w\.-]+@[\w\.-]+\.\w+$")
```

```python
bad_e = data[good_e]
```
