## Data Cleaning and Summary

To begin the analysis, I loaded the dataset into a DataFrame. The data contains **190 rows** and **12 columns**, each representing the performance of paid ads on a monthly basis.

*Here is an overview of each column:*

| Column | Description |
|---|---|
| **Ad Group** | Name of the ad campaign group |
| **Month** | Month the data was recorded |
| **Impressions** | How many times the ad was shown |
| **Clicks** | How many people actually clicked the ad |
| **CTR** | Click-Through Rate (Clicks ÷ Impressions) — measures how compelling the ad is. Higher = better ad copy/targeting |
| **Conversions** | Number of people who clicked AND completed a desired action (purchase, signup, etc.) |
| **Conv Rate** | Conversion Rate = Conversions ÷ Clicks — how well the landing page turns visitors into buyers |
| **Cost** | Total amount spent on the ad group |
| **CPC** | Cost Per Click = Cost ÷ Clicks |
| **Revenue** | Total income generated from the ad group |
| **Sale Amount** | Total value of sales made |
| **P&L** | Profit & Loss = Revenue − Cost. Positive = profitable, negative = loss |

## Descriptive Analysis & Business Observations

The dataset shows large differences in ad performance across all metrics. 
Impressions and clicks vary a lot, with averages of **14,077 impressions** and 
**4,865 clicks**. The high spread shows uneven reach and engagement across campaigns.

The click-through rate averages around **0.27**, with most values between 
**0.19 and 0.36**. This shows moderate engagement but inconsistent results. 
Conversions also differ widely — the median is **70**, while the average is above 
**500**, meaning a few strong ad groups drive most of the results.

The conversion rate is low, averaging **0.08**, with many campaigns falling between 
**0.05 and 0.10**. This shows weak performance in turning traffic into actual results.

Cost and revenue reveal a clear problem. The average cost is **3,344**, while the 
average revenue is **2,957**, leading to an average loss of **386**. Most campaigns 
lose money, with a median loss of around **76** and some losses reaching as low as **−295**.

Only a few campaigns are profitable. The highest profit recorded is **1,507**, while 
the largest loss reaches **−5,672**. Cost per click averages **0.79** and goes up to 
**2.02**. High costs combined with low conversion rates significantly reduce profitability.

> **Overall:** A few campaigns perform well, but most struggle due to poor cost 
> control and low conversion efficiency.

### Deep Exploratory Data Analysis  
1. Campaign Overview: Ad Groups & Performance


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Top and Bottom 10 Ad Groups by Profit
top_profit = data.groupby('Ad Group')['P&L'].sum().sort_values(ascending=False).head(10)
bottom_profit = data.groupby('Ad Group')['P&L'].sum().sort_values(ascending=True).head(10)

# Plot
plt.figure(figsize=(14,5))

plt.subplot(1,2,1)
sns.barplot(x=top_profit.values, y=top_profit.index, hue=top_profit.index, palette='Greens', legend=False)
plt.title('Top 10 Most Profitable Ad Groups')
plt.xlabel('Total Profit ($)')

plt.subplot(1,2,2)
sns.barplot(x=bottom_profit.values, y=bottom_profit.index, hue=bottom_profit.index, palette='Reds_r', legend=False)
plt.title('Bottom 10 Least Profitable Ad Groups')
plt.xlabel('Total Profit/Loss ($)')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

2. Relationship between cost and revenue

```python
sns.scatterplot(data=data, x='Cost', y='Revenue', hue='Ad Group')
plt.title('Cost vs Revenue by Ad Group')
plt.xlabel('Ad Spend ($)')
plt.ylabel('Revenue ($)')
plt.legend([], [], frameon=False)
plt.show()
```
